# Widgets

See the [widgets page of the documentation](https://almond.sh/docs/usage-widgets).

```scala
import almond.widgets._
```

```scala
val slider = IntSlider(value = 5, min = 0, max = 10, description = "n")
val square = Label()
slider.value.observe(n => square.value() = s"n * n = ${n * n}")
VBox(Seq(slider, square))
```

```scala
slider.value() = 7
square.value()
```

```scala
val dropdown = Dropdown(
  options = Seq(1, 2, 3),
  labels = Seq("one", "two", "three"),
  value = Some(2),
  description = "Number:"
)
dropdown
```

```scala
dropdown.value() = Some(3)
(dropdown.index(), dropdown.label())
```

```scala
val button = Button(description = "Click me")
val out    = Output()
button.onClick {
  out.capture(println("clicked"))
}
button.click() // simulates a click
out.outputs().map(_.render)
button
```

```scala
interact(IntSlider(value = 4, max = 10, description = "n")) { n =>
  (1 to n).product
}
```
